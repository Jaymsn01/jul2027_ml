```python
def tower_of_hanoi(disks, src, target, aux):
    if disks == 1:
        print(f'Move {disks} from {src} to {target}')
    elif disks == 2:
        print(f'Move {disks-1} from {src} to {aux}')
        print(f'Move {disks} from {src} to {target}')
        print(f'Move {disks-1} from {aux} to {target}')
    else:
        tower_of_hanoi(disks-1, src, aux, target)
        print(f'Move {disks} from {src} to {target}')
        tower_of_hanoi(disks-1, aux, target, src)

```

```python
tower_of_hanoi(1, 'A', 'C', 'B')
# move 1 from A to C

```

```python
tower_of_hanoi(2, 'A', 'B', 'C')
# move 1 from A to B
# move 2 from A to C
# move 1 from B to C
```
