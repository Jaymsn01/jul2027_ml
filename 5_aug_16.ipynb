```python
def solve2(eqns):
    e1 = eqns[0]
    e2 = eqns[1]
    if e2[0] == 0:
        e1, e2 = e2, e1
    multiplier = e1[0]/e2[0]
    ne_y = [e1[1] - multiplier * e2[1], e1[2]-multiplier*e2[2]]
    y = solve1(ne_y)
    ne_x = [e2[0], e2[2]-y*e2[1]]
    x = solve1(ne_x)
    return [x, y]
```

```python
# # multiplication 
# 4 * 5 -> 4 + 4*4 
# 4 * 4 -> 4 + 4*3

# n * m -> n + n * (m - 1)

def multiply(n, m, tab=''):
    print(f"{tab}Multiplying n:{n}, m:{m}")
    if m == 1:
        return n
    if m == 0:
        return 0
    return n + multiply(n, m - 1, tab + "  ")
```

```python
multiply(2, 5)
```
